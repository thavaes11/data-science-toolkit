```python
#1. Create a numpy array of size 2 x 2

import numpy as np


array_2x2=np.array([[1, 2],
                   [3, 4]])
print("This is my numpy array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4


random_3x4 = np.random.rand(3, 4)

print("This is my random numpy array")
print(random_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

array_of_ones=np.ones([3,4])

array_of_zeros=np.zeros([3,4])

print("This is my array of ones")
print(array_of_ones)

print("This is my array of zeros")
print(array_of_zeros)

```

```python
#4. Create a random numpy array of size 6x3 and transpose it

array_6x3 = np.random.rand(6, 3)

transposed_array = array_6x3.T

print("This is my random array")
print(array_6x3)

print("This is my transposed array")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product


n = 2
p = 3
m = 4

array_nxp = np.random.rand(n, p)
array_pxm= np.random.rand(p, m)

result = np.dot(array_nxp, array_pxm)

print("First array")
print(array_nxp)

print("Second array")
print(array_pxm)

print("Dot product")
print(result)
```
